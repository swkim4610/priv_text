```python
import pickle
import pandas as pd

lastnames = pd.read_csv('lastname.csv', header=None)
firstnames = pd.read_csv('firstname.csv', header=None)
lastnames.columns = ['lastname', 'prob']
firstnames.columns = ['firstname', 'prob']    
lastnames['prob'] = lastnames['prob'].astype(float)
firstnames['prob'] = firstnames['prob'].astype(float)
                     
```

```python
lastnames.info(), firstnames.info()
```

```python
lastname_dict = dict(zip(lastnames['lastname'], lastnames['prob']))
firstname_dict = dict(zip(firstnames['firstname'], firstnames['prob']))
```

```python
with open('lastnames.pkl', 'wb') as file:
    pickle.dump(lastname_dict, file)
with open('firstnames.pkl', 'wb') as file:
    pickle.dump(firstname_dict  , file)
```
